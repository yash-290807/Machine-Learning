```python
features = ["CGPA", "AptitudeTestScore", "CodingTestScore", "MockInterviewScore"]
print(features) #the whole list
print(features[0]) #first iten the indexs start form 0,1,...
print(features[1]) #second item
print(features[-1]) #last item
print(len(features)) #no of otems
#Slicing: feature[start:end]gives you a sub list(end is not oncluded)
print(features[0:2])
#Adding to a list
features.append("Internship")
print(features)
```

Dictinary

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

models = {
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.01),
    "ElasticNet": ElasticNet(alpha=0.01)
}

print(models)                 # Prints the whole dictionary
print(models["Ridge"])        # Access one value by its key

# Looping over a dictionary's keys and values together
for name, model in models.items():
    print(name, ":", model)
```

if/else/ifelse

```python
cgpa = 8.2

if cgpa >= 9:
    tier = "High"
elif cgpa >= 7:
    tier = "Mid"
else:
    tier = "Low"

print(f"CGPA {cgpa} falls into tier: {tier}")
```
